```python
from sympy import *
```

```python
M, F, c, v, q, K0, Kc = symbols('M F c v q K_0 K_c')
v_dot, q_dot = symbols(r'\dot{v} \dot{q}')
theta = symbols('theta')

u = Matrix([q, v])
u
```

```python
u_dot = Matrix([q_dot, v_dot])
u_dot
```

```python
A = Matrix([
    [0, 1],
    [(K0 - theta*Kc)/M, -c/M]
])

B = Matrix([0, F/M])
A
```

```python
B
```

```python
equ1 = Eq(u_dot, A @ u + B)
equ1
```
